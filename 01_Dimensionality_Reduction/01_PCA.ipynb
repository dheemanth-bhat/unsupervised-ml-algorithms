### 1.2 Projection of a Vector

#### Scaler Projection

##### Definition

- Scalar projection just calculates the magnitude of the projection of $x$ vector: $\vec{x}$ on to $y$ vector: $\vec{y}$.
- To get the actual projection vector calculate **Vector Projection**.

$$
\large
\text{Projection of $\vec{x}$ on $\vec{y}$}\;=\;\vec{x}\cdot\widehat{y}
$$

By substituting the formula for unit vector $\widehat{y}$ in above equation we get:

##### Formula

Formula to calculate Scalar projection of $\vec{x}$ on $\vec{y}$:

$$
\large
\text{comp}_{\overrightarrow{y}}\,\vec{x}\;=\;\frac{\vec{x}\cdot\vec{y}}{\Vert{y}\Vert}
$$

#### Vector Projection

##### Definition

The Vector Projection of $\vec{x}$ onto $\vec{y}$ is equal to the Scalar Projection of $\vec{x}$ onto $\vec{y}$ multiplied by the unit vector $\hat{y}$ in the direction of $\vec{y}$.

##### Formula

Formula to calculate Vector projection of $\vec{x}$ on $\vec{y}$:

$$
\large
\text{proj}_{\overrightarrow{y}}\,\vec{x}\;=\;\biggl(\frac{\vec{x}\cdot\vec{y}}{\Vert{y}\Vert}\biggr)\,\widehat{y}
$$

### 1.3 Eigenvector

#### What is Eigenvector?

If a square matrix $S$ exists in such a way that it satisfies the condition: 

$$
\large
S\cdot\vec{u}\;=\;\lambda\cdot\vec{u}
$$

then, $\vec{u}$ is called the **Eigenvector** of **Transformation Matrix** $S$ and $\lambda$ is called as its **Eigenvalue**.

#### Properties

- Eigenvector is a vector which when multiplied with a Transformation Matrix gives back the same vector (in the same direction) but with different magnitude.
- The magnitude can be stretched (positive Eigenvalue) or shrunk (negative Eigenvalue).
- Eigenvector **only gets scaled without changing its direction**.

### 1.4 Mathematical Intuition

#### How PCA Works?

PCA performs two steps to reduce dimensionality of data:

1. Rotating and/or Shifting axis (in the direction of maximum variance).
2. Projection of points

#### How Shifting of Axis works?

- Shifting of axis is indirectly achieved using feature standardization process.
- Features are mean centered i.e., each value is subtracted with the mean value of the same feature.

#### Applications of PCA

1. **Dimensionality reduction**: Reducing the number of features in large datasets to reduce the impact of "curse of dimensionality".
2. **Removes Multicollinearity**: Transforms correlated features into a new set of orthogonal (uncorrelated) axes.
3. **Filtering Noise and Redundancy**: Drops low-variance features that may contain background noise or duplicate information.
4. **Data Visualization**: Visualize higher dimensional data in 2D or 3D.
5. **Faster Model Training**: Reduces model training time.

## 2 Principal Component Analysis (PCA)

### 2.1 Mathematical Intuition

#### STEP #1: Rotating and/or Shifting axis

![PCA Techniques](images/PCA_techniques.png)

- Two features are plotted on a Cartesian plane depicting their correlation on $x$ and $y$ axis.
- The Cartesian plane is rotated/shifted in such a way that the **maximum variance is preserved on one of the axis**.
- New Axes resulting from rotating old axes are called as **Principal Components**.
- In the above example, axis with maximum variance is Principal Component 1 (PC1).
- The other axis automatically perpendicular to Principal Component 1 and is Principal Component 2 (PC2).

#### STEP #2 Projection of points

![PCA Projection of Points](images/PCA_projection_of_points.png)

- Variance on Principal Component 2 will be very minimal.
- Coordinates spread across Principal Component 1 will be **projected** on to that axis.
- Principal Component 2 can now be completely ignored focusing only on Principal Component 1.
- Hence the dimensionality of the data is reduced from 2 to 1 by retaining maximum variance on one of the axis..

> **Note**:
>
> There will be some information loss when projecting points on the Principal Component 1.

### 2.2 Principal Components

#### Properties

- If we have $d$ features we will get maximum $d$ Principal Components.
- Principal Components are **Orthogonal** (perpendicular to each other).
- Principal Components are ordered on importance (Eigen values).
- Principal Components are Linear combination of existing features.

#### Role of Eigenvectors in PCA

- Eigenvalues determine the amount of variance along each Principal Component.
- Eigenvectors represent the direction of these components.

### 2.3 Goal of PCA

#### Geometric Intuition

Goal of PCA:

- After standardizing the feature vector $X$, project $X$ onto a new axis vector $u$.
- Vector $u$ acts as a "new axis" on which the $X$ is projected.
- Goal of PCA is to find that **optimal vector**: $[u_1\,u_2\,\dots\,u_n]$ that **maximizes the variance**.
- In other words maximize the variance in projection of feature vector onto axis vector.

#### PCA Linear Equation

Linear equation for PCA:

$$
\large
\text{feature}_{\text{new}}\;=\;u_1*\text{feature}_1\;+\;u_2*\text{feature}_2\;+\;\ldots\;+\; u_n*\text{feature}_n
$$

Where $\text{feature}_{\text{new}}$ is a linear combination of features: $\{\text{feature}_1,\,\text{feature}_2,\;\ldots,\;\text{feature}_n\}$

### 2.4 Objective Function

#### Variance

Formula of variance:

$$
\large
\sigma^2\;=\;\frac{1}{n}\,\sum_{i\,=\,1}^{n}\,(x_i\;-\;\mu)^2
$$

#### Projection of Vectors

Formula to calculate projection of $\vec{x}$ on $\vec{u}$:

$$
\large
\text{Projection of $\vec{x}$ on $\vec{u}$}\;=\;\frac{x \cdot u}{\Vert{u}\Vert}
$$

#### Optimization Problem

Combining above two equations to formulate an equation for optimization problem of **maximizing variance** during projection of vectors:

$$
\newcommand{\projXU}{\frac{x_i^T \cdot u_i}{\Vert{u}\Vert}}
\large
\vec{u}^*\;=\;\arg\max_{\overrightarrow{u}}\;\frac{1}{n}\,\sum_{i\,=\,1}^n\,\Biggl(\projXU\;-\;\mu\Biggr)^2
$$

Magnitude of axis vector $\vec{u}$ has no impact on optimization problem hence we can set a constraint for it to be equal to 1.

#### Constraint Optimization Problem

##### Constraint

Equation for maximizing variance during projection of vectors with constraint enforced on it:

$$
\newcommand{\projXU}{\frac{x_i^T \cdot u_i}{\Vert{u}\Vert}}
\newcommand{\constraint}{\text{such that}:\;\Vert{\vec{u}}\Vert\,=\,1}
\large
\vec{u}^*\;=\;\arg\max_{\overrightarrow{u}}\;\frac{1}{n}\,\sum_{i\,=\,1}^n\,\Biggl(\projXU\;-\;\mu\Biggr)^2 \quad \constraint
$$

Since the features are standardized $\mu\,=\,0$

#### Objective function for PCA

##### Lagrange Multiplier

Lagrange Multiplier is formulates as:

$$
\large
\mathcal{L}(x,\,\lambda) \equiv f(x)\;+\;\lambda\cdot{g(x)}
$$

Where,

- $f(x)$ is the Objective function
- $\lambda$ is the Lagrange Multiplier
- $g(x)$ is the Constraint function

Hence our Objective Function with constraint can be solved using **Lagrange Multiplier** as:

$$
\newcommand{\dotprod}{\Bigl(x_i^T \cdot u_i\Bigr)^2}
\newcommand{\constraint}{\lambda\bigl(\Vert{\vec{u}}\Vert - 1\bigr)}
\large
\vec{u}^*\;=\;\arg\max_{\overrightarrow{u}}\;\frac{1}{n}\,\sum_{i\,=\,1}^n\,\dotprod\;+\;\constraint
$$

Where $\vec{u}^*$ is the Optimal $\vec{u}$ vector.

#### Gradient Vector

After computing the partial derivatives w.r.t $X$, $u$ and $\lambda$ of Objective function yields:

$$
\large
\bigl(X^T\,X\bigr)\,\vec{u}\;=\;\lambda\,\vec{u}
$$

This is same as Eigenvector equation where,

1. $\vec{u}$ represents the actual Principal Component.
2. $\lambda$ maximum variance captured by that Principal Component.
